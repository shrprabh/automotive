```python
spark.sql("DROP TABLE IF EXISTS automotive_catalog.automotive_mfg.control_metadata")

 

spark.sql("""

    CREATE TABLE automotive_catalog.automotive_mfg.control_metadata (

        table_name        STRING,     -- consuming table that owns the watermark

        last_processed_ts TIMESTAMP,  -- high-watermark of source rows already consumed

        updated_at        TIMESTAMP   -- when this row was last refreshed

    )

    USING DELTA

""")
```

```python
SEED_ROWS = [

    "silver_vehicle_production",

    "silver_quality_inspection",

    "silver_supplier_delivery",

    "gold_vehicle_manufacturing_kpi",

]

 

for table_name in SEED_ROWS:

    spark.sql(f"""

        INSERT INTO automotive_catalog.automotive_mfg.control_metadata

            (table_name, last_processed_ts, updated_at)

        VALUES ('{table_name}', NULL, current_timestamp())

    """)

 

print(f"[SETUP] control_metadata created with {len(SEED_ROWS)} watermark rows")
```
